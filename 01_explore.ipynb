```python
import json 

with open("data/raw/IPL Match Data/335982.json") as f:
    match = json.load(f)
```

```python
print(match.keys())
```

```python
innings = match["innings"]
print(innings[0].keys())
```

```python
first_over = innings[0]["overs"][0]
print(first_over)
```

```python
first_ball = innings[0]["overs"][0]["deliveries"][0]
print(json.dumps(first_ball, indent = 2))
```

```python
runs_scored = 0
balls_faced = 0
sixes = 0
fours = 0

for over in innings[0]["overs"]:
    for delivery in over["deliveries"]:
        if delivery["batter"] == "BB McCullum":

            is_wide = delivery.get("extras",{}).get("wides",0) > 0

            runs_scored += delivery["runs"]["batter"]
            if not is_wide:
                balls_faced += 1
            if delivery["runs"]["batter"] == 6:
                sixes += 1
            if delivery["runs"]["batter"] == 4:
                fours += 1
                
                    
    
strike_rate = (runs_scored / balls_faced)*100

print("Runs Scored by BB McCullum", runs_scored)
print("Balls Faced by BB McCullum", balls_faced)
print("Sixes hit by BB McCullum", sixes)
print("Fours hit by BB McCullum", fours)
print("Strike rate of BB McCullum", f"{strike_rate:.2f}")
```
