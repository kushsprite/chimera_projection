```python
import importlib
import cricket_parser
importlib.reload(cricket_parser)
from cricket_parser import parse_match

import json
import pandas as pd
from pathlib import Path

data_dir = Path("data/raw/IPL Match Data")
match_files = list(data_dir.glob("*.json"))
print(f"Found {len(match_files)} match files")
```

```python
all_player_cards = []
errors = []

for i, filepath in enumerate(match_files):
    try:
        with open(filepath) as f:
            match = json.load(f)
        cards = parse_match(match, match_id=filepath.stem)
        all_player_cards.extend(cards)
    except Exception as e:
        errors.append({"file": filepath.name, "error": str(e)})

    if i % 100 == 0:
        print(f"Processed {i}/{len(match_files)} files...")

print(f"Done. Total player cards: {len(all_player_cards)}")
print(f"Errors: {len(errors)}")
```

```python
print(f"Match Files Found: {len(match_files)}")
print(f"Player Cards Found: {len(all_player_cards)}")
print(f"Average Players per match {len(all_player_cards)/len(match_files):.1f}")
```

```python
df = pd.DataFrame(all_player_cards)
print(df.shape)
print("stumpings" in df.columns)
```
